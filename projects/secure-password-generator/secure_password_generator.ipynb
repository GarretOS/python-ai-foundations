## 🎲 The `random` Module

`import random` loads Python's built-in `random` module so the program can use its random-selection tools. `random.choice(characters)` selects one item from the string stored in `characters`. Each call selects one character from the available pool.

```python
import random

characters = "abcdefghijklmnopqrstuvwxyz"
print(random.choice(characters))
```

## 🧩 Functions, Parameters, and Return Values

A function is a named block of code that can be called when its work is needed. `generate_password()` has four parameters: `length`, `include_numbers`, `include_special`, and `include_uppercase`. A parameter is the name listed in the function definition.

An argument is the actual value supplied when the function is called. In `generate_password(12, True, True, True)`, the values `12`, `True`, `True`, and `True` are arguments matched to the four parameters.

`return` sends a value back to the code that called the function. This is different from `print()`: `print()` displays a value, while `return` makes a value available for later use.

## 🔐 Password Generation Algorithm

The generator follows these steps:

1. Start with lowercase letters.
2. Add numbers if the user selected them.
3. Add special characters if selected.
4. Add uppercase letters if selected.
5. Start with an empty password.
6. Choose one random character for each requested position.
7. Add each chosen character to the password and return the finished string.

```python
def generate_password(length, include_numbers, include_special, include_uppercase):
    characters = "abcdefghijklmnopqrstuvwxyz"

    if include_numbers:
        characters += "0123456789"

    if include_special:
        characters += "!@#$%^&*"

    if include_uppercase:
        characters += "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

    password = ""

    for _ in range(length):
        password += random.choice(characters)

    return password
```

## 🔁 `for` Loops and `range()`

`for _ in range(length):` repeats the loop once for every requested password position. `range(5)` represents the sequence `0, 1, 2, 3, 4`, which contains five values and therefore causes five iterations. The ending number is excluded, but the count is still five.

The underscore `_` is a conventional variable name used when the loop value itself is not needed. The generator only needs to repeat the action; it does not need the current iteration number.

## ⚠️ Input and Error Handling

`input()` reads text typed by the user. Because it returns a string, the program uses `int()` to convert the password length into an integer. Text such as `abc` or `12.5` cannot be converted by `int()` and raises a `ValueError`.

The `try` block contains the conversion that might fail. `except ValueError` handles that specific problem and asks for the length again.

`continue` skips the rest of the current loop iteration and starts the next one. It is used when the length is below 8.

The `get_yes_no(prompt)` helper uses `while True` to keep asking a categorical question until the answer is valid. It normalizes input with `.lower()`, accepts `yes` or `y` with `return True`, and accepts `no` or `n` with `return False`. Any other response, including empty input, displays a friendly validation message and retries. This prevents invalid text from being silently interpreted as `False`.

`break` exits the main loop after a valid length and the character choices have been collected.

## 💪 Password Strength

This project uses a simple educational scoring rule. The starting score is the password length. Numbers add 2 points, special characters add 3 points, and uppercase letters add 2 points. Scores below 10 are `Weak`, scores below 16 are `Medium`, and all other scores are `Strong`.

This is not a professional cybersecurity measurement. It is a small example for practicing `len()`, Boolean conditions, arithmetic, and `if` / `elif` / `else`.

```python
def check_password_strength(password, include_numbers, include_special, include_uppercase):
    score = len(password)

    if include_numbers:
        score += 2

    if include_special:
        score += 3

    if include_uppercase:
        score += 2

    if score < 10:
        return "Weak"
    elif score < 16:
        return "Medium"
    else:
        return "Strong"
```

## 🧪 Mini Unit Test

A unit test is a small check of one part of a program. This project test calls `generate_password()` with a requested length of 12 and checks only whether the returned password has 12 characters. It reports the result with `print()` and a normal `if` / `else`.

One passing test does not prove that the entire program is correct. It does not check every character option, input error, or interactive path.

```python
def test_generate_password():
    test_password = generate_password(12, True, True, True)

    if len(test_password) == 12:
        print("Test passed: generated password has the correct length.")
    else:
        print("Test failed: generated password has the wrong length.")


def get_yes_no(prompt):
    while True:
        answer = input(prompt).lower()

        if answer == "yes" or answer == "y":
            return True
        elif answer == "no" or answer == "n":
            return False
        else:
            print("Please enter yes/y or no/n.")
```

## ▶️ Run the Password Generator

The next cell contains the complete project code. Run it and enter a length of at least 8, followed by `yes` / `y` or `no` / `n` for each character group. Invalid responses are rejected and the same question is asked again. The `get_yes_no(prompt)` helper returns a Boolean for each valid choice. The `if __name__ == "__main__":` guard starts `main()` when the file or cell is run directly, but prevents `main()` from starting automatically when the module is imported.

```python
import random


def generate_password(length, include_numbers, include_special, include_uppercase):
    characters = "abcdefghijklmnopqrstuvwxyz"

    if include_numbers:
        characters += "0123456789"

    if include_special:
        characters += "!@#$%^&*"

    if include_uppercase:
        characters += "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

    password = ""

    for _ in range(length):
        password += random.choice(characters)

    return password


def check_password_strength(password, include_numbers, include_special, include_uppercase):
    score = len(password)

    if include_numbers:
        score += 2

    if include_special:
        score += 3

    if include_uppercase:
        score += 2

    if score < 10:
        return "Weak"
    elif score < 16:
        return "Medium"
    else:
        return "Strong"


def test_generate_password():
    test_password = generate_password(12, True, True, True)

    if len(test_password) == 12:
        print("Test passed: generated password has the correct length.")
    else:
        print("Test failed: generated password has the wrong length.")


def get_yes_no(prompt):
    while True:
        answer = input(prompt).lower()

        if answer == "yes" or answer == "y":
            return True
        elif answer == "no" or answer == "n":
            return False
        else:
            print("Please enter yes/y or no/n.")


def main():
    while True:
        try:
            length_input = input("Enter password length (minimum 8): ")
            length = int(length_input)

            if length < 8:
                print("The password must be at least 8 characters long.")
                continue

            include_numbers = get_yes_no("Include numbers? (yes/no): ")
            include_special = get_yes_no(
                "Include special characters? (yes/no): "
            )
            include_uppercase = get_yes_no(
                "Include uppercase letters? (yes/no): "
            )

            new_password = generate_password(
                length,
                include_numbers,
                include_special,
                include_uppercase,
            )
            strength = check_password_strength(
                new_password,
                include_numbers,
                include_special,
                include_uppercase,
            )

            print(f"Generated password: {new_password}")
            print(f"Password strength: {strength}")

            break

        except ValueError:
            print("Please enter a whole number for the password length.")


if __name__ == "__main__":
    main()
```
