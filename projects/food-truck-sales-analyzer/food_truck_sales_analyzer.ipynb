# 🍔 Food Truck Sales Analyzer

## 🎯 Project Overview

Stack Attack is a fictional burger food truck. This notebook analyzes one week of sales to find revenue totals, busy days, burger targets, and simple sales comparisons. The project stays within the scope of an introductory Python lesson.

## 🐍 Python Concepts

This project uses dictionaries, lists, variables, `for` and `while` loops, conditionals, counters, accumulators, list indexing, `len()`, arithmetic, comparison operators, and f-strings.

## 📦 Data Setup

The weekly sales data is stored in a dictionary with one list for each type of information.

```python
# Weekly sales data for the fictional Stack Attack burger truck
food_truck_sales = {
    "day": ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"],
    "burgers_sold": [46, 53, 49, 61, 74, 96, 83],
    "drinks_sold": [31, 40, 34, 55, 47, 71, 62],
    "revenue": [520, 610, 565, 850, 760, 1115, 955]
}

day_list = food_truck_sales["day"]
burger_list = food_truck_sales["burgers_sold"]
drink_list = food_truck_sales["drinks_sold"]
revenue_list = food_truck_sales["revenue"]
```

## 💰 Average Daily Revenue

Add each day's revenue and divide by the number of days.

```python
total_revenue = 0
for amount in revenue_list:
    total_revenue = total_revenue + amount
average_revenue = total_revenue / len(revenue_list)
print(f"Average daily revenue: ${average_revenue:.2f}")
```

The average daily revenue is **\$767.86**.

## 🏆 Busiest Burger-Sales Day

Use a running maximum and list indexes to connect the largest burger count to its day.

```python
max_burgers = 0
max_day = ""
index = 0
while index < len(burger_list):
    if burger_list[index] > max_burgers:
        max_burgers = burger_list[index]
        max_day = day_list[index]
    index = index + 1
print(f"Busiest burger-sales day: {max_day} with {max_burgers} burgers sold")
```

Saturday was the busiest burger-sales day, with **96 burgers sold**.

## 📈 Total Weekly Revenue

The weekly total is the accumulator from the revenue calculation.

```python
print(f"Total weekly revenue: ${total_revenue:,.2f}")
```

Stack Attack earned **\$5,375.00** during the week.

## 🔎 Simple Sales and Revenue Relationship Check

Calculate average burger sales, then count above-average burger days and check their revenue. This is a simple comparison, not statistical correlation.

```python
total_burgers = 0
for burgers in burger_list:
    total_burgers = total_burgers + burgers
average_burgers = total_burgers / len(burger_list)

days_burgers_above_average = 0
days_burgers_and_revenue_above_average = 0
index = 0
while index < len(burger_list):
    if burger_list[index] > average_burgers:
        days_burgers_above_average = days_burgers_above_average + 1
        if revenue_list[index] > average_revenue:
            days_burgers_and_revenue_above_average = days_burgers_and_revenue_above_average + 1
    index = index + 1

if days_burgers_above_average > 0:
    relationship_ratio = days_burgers_and_revenue_above_average / days_burgers_above_average
else:
    relationship_ratio = 0

print(f"Average burger sales: {average_burgers:.1f}")
print(f"Above-average burger days also having above-average revenue: {days_burgers_and_revenue_above_average} out of {days_burgers_above_average}")
print(f"Simple relationship ratio: {relationship_ratio:.2f}")
```

There were **2 out of 3** above-average burger days with above-average revenue. The simple ratio is **0.67**; this does not measure statistical correlation.

## 📅 Weekday vs. Weekend Burger Sales

Use separate totals and counters for weekdays and the Saturday-Sunday weekend.

```python
weekday_total = 0
weekday_count = 0
weekend_total = 0
weekend_count = 0
index = 0
while index < len(day_list):
    if day_list[index] == "Saturday" or day_list[index] == "Sunday":
        weekend_total = weekend_total + burger_list[index]
        weekend_count = weekend_count + 1
    else:
        weekday_total = weekday_total + burger_list[index]
        weekday_count = weekday_count + 1
    index = index + 1
average_weekday_burgers = weekday_total / weekday_count
average_weekend_burgers = weekend_total / weekend_count
print(f"Average weekday burger sales: {average_weekday_burgers:.1f}")
print(f"Average weekend burger sales: {average_weekend_burgers:.1f}")
```

Average weekday burger sales were **56.6**, compared with **89.5** on weekends.

## 🎯 Burger Sales Target

Count every day meeting or exceeding the 70-burger target. The comparison uses `>=`.

```python
threshold = 70
days_meeting_target = 0
for burgers in burger_list:
    if burgers >= threshold:
        days_meeting_target = days_meeting_target + 1
print(f"Days meeting the {threshold}-burger target: {days_meeting_target}")
```

**3 days** met or exceeded the burger-sales target.

## 🥤 Burgers vs. Drinks

Compare the total number of burgers and drinks sold. This is a small extension using concepts already learned.

```python
total_drinks = 0
for drinks in drink_list:
    total_drinks = total_drinks + drinks

if total_burgers > total_drinks:
    higher_selling_category = "burgers"
elif total_drinks > total_burgers:
    higher_selling_category = "drinks"
else:
    higher_selling_category = "both categories equally"

print(f"Total burgers sold: {total_burgers}")
print(f"Total drinks sold: {total_drinks}")
print(f"Higher-selling category: {higher_selling_category}")
```

Stack Attack sold **462 burgers** and **340 drinks**, so burgers were the higher-selling category.
